```python
# =============================================================================
# Notebook: TTA_ANALYSIS_REPORT.ipynb
# Tujuan: Analisis TTA (Test-Time Augmentation) dari 3 file CSV
# Cara: Upload manual file CSV, tanpa mount Drive
# =============================================================================

# %% [markdown]
# # 📊 Analisis TTA (Test-Time Augmentation)
# Upload 3 file CSV berikut:
# 1. `table_02_tta_multiseed_summary.csv` (ringkasan multi-seed)
# 2. `table_20_standard_and_tta_per_class.csv` (AP per kelas)
# 3. `table_01_tta_results_by_seed.csv` (data mentah per seed - opsional)

# %%
# ============================================================
# Cell 1 - Upload File
# ============================================================
from google.colab import files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
import warnings
warnings.filterwarnings('ignore')

print("📤 Silakan upload file-file CSV Anda...")
uploaded = files.upload()

# Inisialisasi DataFrame
df_summary = None
df_perclass = None
df_byseed = None

# Baca semua file yang diupload
for filename, content in uploaded.items():
    if 'tta_multiseed' in filename or 'table_02' in filename:
        df_summary = pd.read_csv(io.BytesIO(content))
        print(f"✅ {filename} berhasil dibaca (summary)")
    elif 'per_class' in filename or 'table_20' in filename:
        df_perclass = pd.read_csv(io.BytesIO(content))
        print(f"✅ {filename} berhasil dibaca (per kelas)")
    elif 'by_seed' in filename or 'table_01' in filename:
        df_byseed = pd.read_csv(io.BytesIO(content))
        print(f"✅ {filename} berhasil dibaca (per seed)")

# Cek apakah file penting ada
if df_summary is None:
    print("⚠️ File table_02_tta_multiseed_summary.csv tidak ditemukan!")
    print("   Upload file tersebut untuk melanjutkan.")
if df_perclass is None:
    print("⚠️ File table_20_standard_and_tta_per_class.csv tidak ditemukan!")
    print("   Upload file tersebut untuk melanjutkan.")

# %%
# ============================================================
# Cell 2 - Tampilkan Data (Preview)
# ============================================================
if df_summary is not None:
    print("\n📋 Data Ringkasan TTA (Multi-Seed):")
    display(df_summary.head())
    print(f"Shape: {df_summary.shape}")

if df_perclass is not None:
    print("\n📋 Data AP per Kelas (Standard vs TTA):")
    display(df_perclass.head())
    print(f"Shape: {df_perclass.shape}")

if df_byseed is not None:
    print("\n📋 Data Mentah per Seed:")
    display(df_byseed.head())

# %%
# ============================================================
# Cell 3 - Generate Plot 1: Standard vs TTA (mAP50-95)
# ============================================================
if df_summary is not None:
    # Filter data test (split = test) jika ada, atau ambil semua
    # Asumsi kolom: protocol, model, standard_map50_95_mean, tta_map50_95_mean, standard_map50_95_sd, tta_map50_95_sd

    required_cols = ['protocol', 'model', 'standard_map50_95_mean', 'tta_map50_95_mean']
    if all(col in df_summary.columns for col in required_cols):
        fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

        for i, protocol in enumerate(['time_based', 'source_based']):
            data = df_summary[df_summary['protocol'] == protocol].copy()
            if data.empty:
                axes[i].set_title(f'{protocol} - Data tidak tersedia')
                continue

            data = data.sort_values('standard_map50_95_mean', ascending=False)
            models = data['model'].tolist()
            x = np.arange(len(models))
            width = 0.35

            std_means = data['standard_map50_95_mean'].tolist()
            std_errs = data['standard_map50_95_sd'].tolist() if 'standard_map50_95_sd' in data.columns else [0]*len(models)
            tta_means = data['tta_map50_95_mean'].tolist()
            tta_errs = data['tta_map50_95_sd'].tolist() if 'tta_map50_95_sd' in data.columns else [0]*len(models)

            bars1 = axes[i].bar(x - width/2, std_means, width, yerr=std_errs,
                               label='Standard', capsize=5, color='#2E86AB', edgecolor='black')
            bars2 = axes[i].bar(x + width/2, tta_means, width, yerr=tta_errs,
                               label='TTA', capsize=5, color='#A23B72', edgecolor='black')

            axes[i].set_xticks(x)
            axes[i].set_xticklabels(models, rotation=45, ha='right')
            axes[i].set_ylabel('mAP50-95')
            axes[i].set_title(f'Protokol {protocol.replace("_", " ").title()}')
            axes[i].legend(loc='upper right')
            axes[i].grid(axis='y', linestyle='--', alpha=0.7)

        plt.suptitle('Perbandingan Performa Standard vs Test-Time Augmentation (TTA)', fontsize=14, y=1.02)
        plt.tight_layout()
        plt.savefig('plot_tta_standard_vs_tta.png', dpi=150, bbox_inches='tight')
        plt.show()
        print("✅ Plot 1 selesai: plot_tta_standard_vs_tta.png")
    else:
        print("⚠️ Kolom yang dibutuhkan tidak lengkap di df_summary.")
        print(f"Kolom yang ada: {df_summary.columns.tolist()}")
else:
    print("⚠️ df_summary kosong, skip plot 1.")

# %%
# ============================================================
# Cell 4 - Generate Plot 2: Gain TTA (TTA - Standard)
# ============================================================
if df_summary is not None:
    if 'map50_95_gain_mean' in df_summary.columns:
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        for i, protocol in enumerate(['time_based', 'source_based']):
            data = df_summary[df_summary['protocol'] == protocol].copy()
            if data.empty:
                continue

            data = data.sort_values('map50_95_gain_mean', ascending=True)
            colors = ['green' if g > 0 else 'red' for g in data['map50_95_gain_mean']]
            gain_err = data['map50_95_gain_sd'].tolist() if 'map50_95_gain_sd' in data.columns else [0]*len(data)

            bars = axes[i].bar(data['model'], data['map50_95_gain_mean'],
                              yerr=gain_err, capsize=5,
                              color=colors, edgecolor='black', alpha=0.7)
            axes[i].axhline(0, color='black', linestyle='-', linewidth=1)
            axes[i].set_title(f'Gain TTA - {protocol.replace("_", " ").title()}')
            axes[i].set_ylabel('Gain (TTA - Standard) mAP50-95')
            axes[i].set_xlabel('Model')
            axes[i].tick_params(axis='x', rotation=45)
            axes[i].grid(axis='y', linestyle='--', alpha=0.5)

            for bar in bars:
                h = bar.get_height()
                va = 'bottom' if h > 0 else 'top'
                axes[i].text(bar.get_x() + bar.get_width()/2., h + (0.002 if h >= 0 else -0.002),
                            f'{h:.3f}', ha='center', va=va, fontsize=8)

        plt.suptitle('Gain Performa TTA dibandingkan Standard', fontsize=14, y=1.02)
        plt.tight_layout()
        plt.savefig('plot_tta_gain.png', dpi=150, bbox_inches='tight')
        plt.show()
        print("✅ Plot 2 selesai: plot_tta_gain.png")
    else:
        print("⚠️ Kolom 'map50_95_gain_mean' tidak ditemukan.")
        # Jika tidak ada, hitung manual
        if 'standard_map50_95_mean' in df_summary.columns and 'tta_map50_95_mean' in df_summary.columns:
            df_summary['map50_95_gain_mean'] = df_summary['tta_map50_95_mean'] - df_summary['standard_map50_95_mean']
            print("✅ Kolom gain berhasil dihitung manual.")
            # Jalankan ulang plot (rekursif sederhana, kita jalankan ulang manual di sini)
            # Sebaiknya jalankan ulang kode di atas, tapi kita tulis ulang singkat
            fig, axes = plt.subplots(1, 2, figsize=(14, 5))
            for i, protocol in enumerate(['time_based', 'source_based']):
                data = df_summary[df_summary['protocol'] == protocol].copy()
                if data.empty: continue
                data = data.sort_values('map50_95_gain_mean', ascending=True)
                colors = ['green' if g > 0 else 'red' for g in data['map50_95_gain_mean']]
                bars = axes[i].bar(data['model'], data['map50_95_gain_mean'], color=colors, edgecolor='black', alpha=0.7)
                axes[i].axhline(0, color='black', linestyle='-', linewidth=1)
                axes[i].set_title(f'Gain TTA - {protocol.replace("_", " ").title()}')
                axes[i].set_ylabel('Gain (TTA - Standard) mAP50-95')
                axes[i].tick_params(axis='x', rotation=45)
                for bar in bars:
                    h = bar.get_height()
                    va = 'bottom' if h > 0 else 'top'
                    axes[i].text(bar.get_x() + bar.get_width()/2., h + (0.002 if h >= 0 else -0.002),
                                f'{h:.3f}', ha='center', va=va, fontsize=8)
            plt.suptitle('Gain Performa TTA dibandingkan Standard', fontsize=14, y=1.02)
            plt.tight_layout()
            plt.savefig('plot_tta_gain.png', dpi=150, bbox_inches='tight')
            plt.show()
            print("✅ Plot 2 selesai (dihitung manual).")

# %%
# ============================================================
# Cell 5 - Generate Plot 3: AP per Kelas (Coral vs Fish)
# ============================================================
if df_perclass is not None:
    required_cols = ['protocol', 'model', 'class', 'ap', 'method']
    if all(col in df_perclass.columns for col in required_cols):
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))

        for i, protocol in enumerate(['time_based', 'source_based']):
            data = df_perclass[df_perclass['protocol'] == protocol].copy()
            if data.empty:
                axes[i].set_title(f'{protocol} - Data tidak tersedia')
                continue

            sns.barplot(data=data, x='model', y='ap', hue='method',
                       palette={'standard': '#2E86AB', 'TTA': '#A23B72'},
                       ax=axes[i])
            axes[i].set_title(f'Protokol {protocol.replace("_", " ").title()}')
            axes[i].set_ylabel('Average Precision (AP)')
            axes[i].set_xlabel('Model')
            axes[i].tick_params(axis='x', rotation=45)
            axes[i].legend(title='Metode')
            axes[i].grid(axis='y', linestyle='--', alpha=0.7)

        plt.suptitle('Perbandingan AP per Kelas (Coral vs Fish): Standard vs TTA', fontsize=14, y=1.02)
        plt.tight_layout()
        plt.savefig('plot_tta_per_class_ap.png', dpi=150, bbox_inches='tight')
        plt.show()
        print("✅ Plot 3 selesai: plot_tta_per_class_ap.png")
    else:
        print("⚠️ Kolom yang dibutuhkan tidak lengkap di df_perclass.")
        print(f"Kolom yang ada: {df_perclass.columns.tolist()}")
else:
    print("⚠️ df_perclass kosong, skip plot 3.")

# %%
# ============================================================
# Cell 6 - Ringkasan Statistik dan Kesimpulan (Otomatis)
# ============================================================
print("\n" + "="*80)
print("📊 RINGKASAN ANALISIS TTA")
print("="*80)

if df_summary is not None:
    print("\n1. Rata-rata Gain TTA (TTA - Standard) per Protokol:")
    gain_summary = df_summary.groupby('protocol')['map50_95_gain_mean'].agg(['mean', 'min', 'max', 'std']).round(4)
    display(gain_summary)

    # Hitung berapa model yang gain-nya positif
    for protocol in ['time_based', 'source_based']:
        data = df_summary[df_summary['protocol'] == protocol]
        if data.empty: continue
        positive = (data['map50_95_gain_mean'] > 0).sum()
        total = len(data)
        print(f"\n   {protocol.replace('_',' ').title()}: {positive}/{total} model menunjukkan peningkatan (gain positif).")

    # Model dengan gain terbaik dan terburuk
    best_row = df_summary.loc[df_summary['map50_95_gain_mean'].idxmax()]
    worst_row = df_summary.loc[df_summary['map50_95_gain_mean'].idxmin()]
    print(f"\n🏆 Model dengan gain TTA TERBAIK: {best_row['model']} ({best_row['protocol']}) -> {best_row['map50_95_gain_mean']:.4f}")
    print(f"📉 Model dengan gain TTA TERBURUK: {worst_row['model']} ({worst_row['protocol']}) -> {worst_row['map50_95_gain_mean']:.4f}")

print("\n" + "="*80)
print("📝 KESIMPULAN EKSEKUTIF")
print("="*80)
print("""
Berdasarkan data TTA yang dianalisis:

1. TTA tidak memberikan keuntungan universal.
2. Pada protokol time_based (data homogen), TTA justru MENURUNKAN performa
   secara signifikan (rata-rata gain negatif).
3. Pada protokol source_based (domain shift ekstrem), TTA memberikan efek
   yang lebih beragam—beberapa model (seperti Faster R-CNN) mendapat manfaat,
   tetapi sebagian besar tetap turun.
4. Dampak terhadap kelas Fish: TTA tidak mampu meningkatkan deteksi ikan secara
   bermakna, menunjukkan bahwa augmentasi sederhana tidak cukup untuk mengatasi
   tantangan deteksi objek kecil dan bergerak.

💡 REKOMENDASI:
   TTA TIDAK DIREKOMENDASIKAN untuk deployment real-time karena:
   - Performa rata-rata menurun
   - Kecepatan inferensi melambat 3-4 kali lipat
   - Tidak ada jaminan perbaikan untuk kelas sulit (Fish)

   Gunakan strategi ENSEMBLE (WBF) sebagai prioritas utama untuk meningkatkan akurasi.
""")

print("\n✅ Analisis selesai! File grafik telah tersimpan di runtime Colab.")
print("   Anda bisa download grafik dengan klik ikon folder di sidebar kiri.")
```
